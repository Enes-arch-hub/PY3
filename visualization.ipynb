```python
import pandas as pd
import matplotlib.pyplot as plt

students = {
    "Name": ["Enes", "Alice", "Brian", "Faith", "Mercy"],
    "Score": [88, 95, 79, 91, 85]
}

df = pd.DataFrame(students)

plt.figure(figsize=(7,4))
plt.bar(df["Name"], df["Score"])

plt.title("Student Scores")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```
