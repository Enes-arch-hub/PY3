```python
import pandas as pd
import matplotlib.pyplot as plt
#1. Create the football players dataset
players = {
    "Player": ["Lionel Messi", "Cristiano Ronaldo", "Neymar Jr", "Kylian Mbappe", "Kevin De Bruyne"],
    "Goals": [30, 25, 20, 22, 18],
    "Assists": [12, 10, 15, 14, 16],
}
#2. Create a Dataframe
df = pd.DataFrame(players)

#3. Calculate the total contributions (Goals + Assists) for each player
df["Total Contributions"] = df["Goals"] + df["Assists"]
# Display the complete DataFrame
print("Football Players Statistics:")
print("-------------------------------")
print(df)

#4. Find the player with the highest total contributions
top_player = df.loc[df["Total Contributions"].idxmax()]
print("\nPlayer with the highest total contributions:")
print("---------------------------------------------")
print("Player:", top_player["Player"])
print("Goals:", top_player["Goals"])
print("Assists:", top_player["Assists"])
print("Total Contributions:", top_player["Total Contributions"])

#5.Create a bar chart to visualize the total contributions of each player
plt.figure(figsize=(8, 5))
plt.bar(df["Player"], df["Total Contributions"])
plt.xlabel("Player")
plt.ylabel("Total Contributions")
plt.title("Football Players Total Contributions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
